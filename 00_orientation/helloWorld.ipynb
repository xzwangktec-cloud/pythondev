見出し

見出しは「#」を使って指定します。#の数で見出しの階層を決めます。

# 見出し1
## 見出し2
### 見出し3
強調

太字：**太字** または __太字__
斜体：*斜体* または _斜体_
太字かつ斜体：***太字かつ斜体***

水平線

3つ以上のハイフン、アスタリスク、またはアンダースコアで水平線が作成できます。

---
テーブル

パイプ（|）とハイフン（-）でテーブルを作成します。

| 見出し1 | 見出し2 |
| ------- | ------- |
| データ1 | データ2 |
| データ3 | データ4 |
